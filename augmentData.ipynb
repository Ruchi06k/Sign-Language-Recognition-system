# Augment

```python
import cv2
import numpy as np
import os
import random

def augment_and_save(image_path, save_dir, augment_count=3):
    img = cv2.imread(image_path)
    img_name = os.path.splitext(os.path.basename(image_path))[0]

    for i in range(augment_count):
        aug_img = img.copy()

        # Rotate
        angle = random.randint(-15, 15)
        M = cv2.getRotationMatrix2D((aug_img.shape[1] // 2, aug_img.shape[0] // 2), angle, 1)
        aug_img = cv2.warpAffine(aug_img, M, (aug_img.shape[1], aug_img.shape[0]))

        # Brightness variation
        hsv = cv2.cvtColor(aug_img, cv2.COLOR_BGR2HSV)
        hsv[:, :, 2] = np.clip(hsv[:, :, 2] * random.uniform(0.8, 1.2), 0, 255)
        aug_img = cv2.cvtColor(hsv, cv2.COLOR_HSV2BGR)

        # Horizontal flip randomly
        if random.choice([True, False]):
            aug_img = cv2.flip(aug_img, 1)

        # Save the augmented image
        save_path = os.path.join(save_dir, f"{img_name}_aug{i}.png")
        cv2.imwrite(save_path, aug_img)

# === Main block ===

base_dir = 'Image'  # Folder containing A-Z folders

# Loop through all folders A–Z
for label in os.listdir(base_dir):
    folder_path = os.path.join(base_dir, label)
    if os.path.isdir(folder_path):
        print(f"Augmenting images in: {label}")
        for img_file in os.listdir(folder_path):
            img_path = os.path.join(folder_path, img_file)
            try:
                augment_and_save(img_path, folder_path, augment_count=3)  # 3 new images per original
            except Exception as e:
                print(f"Error with {img_path}: {e}")
```
