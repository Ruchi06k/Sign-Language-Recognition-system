```python
import os
import cv2

# Initializing webcam
cap = cv2.VideoCapture(0)
directory = 'Image/'

# Creating folders if they don't exist
for letter in [chr(i) for i in range(65, 91)]:
    path = os.path.join(directory, letter)
    os.makedirs(path, exist_ok=True)

while True:
    _, frame = cap.read()
    count = {chr(i): len(os.listdir(os.path.join(directory, chr(i)))) for i in range(65, 91)}

    # Draw ROI box
    cv2.rectangle(frame, (0, 40), (300, 400), (255, 255, 255), 2)

    # To display counts on the right (outside ROI)
    y = 100
    for key in sorted(count.keys()):
        text = f"{key} : {count[key]}"
        cv2.putText(frame, text, (320, y), cv2.FONT_HERSHEY_PLAIN, 1, (0, 255, 255), 1)
        y += 10

    # Showing the live frame and ROI preview
    cv2.imshow("Live Feed", frame)
    cv2.imshow("ROI", frame[40:400, 0:300])

    # Crop only ROI for saving
    roi = frame[40:400, 0:300]

    key = cv2.waitKey(10) & 0xFF

    if key == 27:  # ESC to quit
        break
    elif 97 <= key <= 122:  # keys 'a' to 'z'
        letter = chr(key).upper()
        if letter in count:
            path = os.path.join(directory, letter, f"{count[letter]}.png")
            cv2.imwrite(path, roi)
            print(f"Saved: {path}")


cap.release()
cv2.destroyAllWindows()
```
